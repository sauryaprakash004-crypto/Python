```python
import numpy as np
import pandas as pd
```

```python
h=pd.read_csv('movies.csv')
h.head(1)
```

```python
k=h.assign(genres=h['genres'].str.split('|')).explode('genres')
p=k['genres'].value_counts().head(3).index.tolist()
avg_ratings=h[h['genres'].isin(p)].groupby('genres')['imdb_rating'].mean()
print(f"top 3 genres:{p}")
print(avg_ratings)
```
