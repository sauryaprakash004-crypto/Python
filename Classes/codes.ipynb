```python


def rev(a,b,c):
    leftpointer=b
    rightpointer=c-1
    while(rightpointer>leftpointer):
        temp=a[rightpointer]
        a[rightpointer]=a[leftpointer]
        a[leftpointer]=temp
        leftpointer+=1
        rightpointer-=1
    return a

a=[1,2,3,4,5]
b=0
c=2
rev(a,b,c)
rev(a,2,5)
rev(a,0,5)
print(a)






```

```python
k=d[d['over']>15].copy()
n=k.drop_duplicates(['over','match_id'])
g=n.value_counts('bowler').reset_index()
ll=k.groupby('bowler').agg(a=('bye_runs','sum'),b=('legbye_runs','sum'),c=('total_runs','sum'))
ll['tr']=ll['c']-ll['b']-ll['a']
kp=g.merge(ll,on='bowler')
kp['economy']=kp['tr']/kp['count']
kp=kp[kp['count']>20]
dd=kp.loc[:,['bowler','economy']]
dd.sort_values(by='economy').head(5)
```

```python
# 1. Filter for Death Overs
k = d[d['over'] > 15].copy()

# 2. Define 'Legal Ball' (Not a wide, not a no-ball)
k['is_legal_ball'] = (k['wide_runs'] == 0) & (k['noball_runs'] == 0)

# 3. Aggregate runs and legal balls
stats = k.groupby('bowler').agg(
    off_runs=('total_runs', 'sum'),
    byes=('bye_runs', 'sum'),
    legbyes=('legbye_runs', 'sum'),
    legal_balls=('is_legal_ball', 'sum')
)

# 4. Bowler's Runs (Exclude byes and leg-byes)
stats['bowler_runs'] = stats['off_runs'] - stats['byes'] - stats['legbyes']

# 5. Convert balls to overs (Balls / 6)
stats['overs_bowled'] = stats['legal_balls'] / 6

# 6. Filter for threshold (e.g., more than 20 overs)
stats = stats[stats['overs_bowled'] >= 20]

# 7. Calculate Economy
stats['economy'] = stats['bowler_runs'] / stats['overs_bowled']

# Result
print(stats['economy'].sort_values().head(5))
```
